```python
import streamlit as st
import pandas as pd
import geopandas as gpd
from streamlit_folium import st_folium
from rainfall_eda import interactive_precip_histogram, perform_ks_test, create_interactive_table, map_ks_test_results, generate_basic_plot

# Load data
try:
    zambia_rain_df = pd.read_csv("../data/zambia_admin2_rs_precip.csv")
    admin_boundaries = gpd.read_file("../data/zambia_admin2_boundaries.geojson")
except FileNotFoundError as e:
    st.error(f"File not found: {e}")
    st.stop()

# Merge geometry
zambia_rain_df = gpd.GeoDataFrame(zambia_rain_df.merge(admin_boundaries, on="admin2_name"))

# Perform KS Test and append results
results_df = perform_ks_test(zambia_rain_df)
zambia_rain_df = zambia_rain_df.merge(results_df, on="admin2_name")

# Streamlit Tabs
tab1, tab2 = st.tabs(["Main Analysis", "Additional Plots"])

# Tab 1
with tab1:
    st.sidebar.title("Select Analysis Type")
    analysis_type = st.sidebar.selectbox("Options", ["Precipitation Histogram", "Kolmogorov-Smirnov Test", "Map KS Test Results"])
    if analysis_type == "Precipitation Histogram":
        st.subheader("Precipitation Histogram")
        fig = interactive_precip_histogram(zambia_rain_df, "admin2_name", "precipitation")
        st.plotly_chart(fig, use_container_width=True)
    elif analysis_type == "Kolmogorov-Smirnov Test":
        st.subheader("Kolmogorov-Smirnov Test")
        table_fig = create_interactive_table(results_df)
        st.plotly_chart(table_fig, use_container_width=True)
    elif analysis_type == "Map KS Test Results":
        st.subheader("KS Test Results Map")
        folium_map = map_ks_test_results(zambia_rain_df)
        st_folium(folium_map, width=700, height=500)

# Tab 2
with tab2:
    st.subheader("Additional Analysis")
    fig = generate_basic_plot(zambia_rain_df)
    st.plotly_chart(fig, use_container_width=True)
```
